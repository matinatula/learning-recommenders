````python
# Function to convert all non-integer IDs to NaN

def clean_ids(x):
    try:
        return int(x)
    except:
        return np.nan
````

````python
# Clean the ids of df
df['id'] = df['id'].apply(clean_ids)

# Filter all rows that have a null ID
df = df[df['id'].notnull()]
````

````python
# Convert the IDs into integer
df['id'] = df['id'].astype('int')
key_df['id'] = key_df['id'].astype('int')
cred_df['id'] = cred_df['id'].astype('int')

# Merge keywords and credits into your main metatdata dataframe
df = df.merge(cred_df, on='id')
df = df.merge(key_df, on='id')

# Display the head of df
df.head()
````

````python
# Convert the stringified objects into the native python objects

from ast import literal_eval

features = ['cast', 'crew', 'keywords', 'genres']
for feature in features:
    df[feature] = df[feature].apply(literal_eval)
````

````python
# Extract the director's name. If director is not listed, return NaN

def get_director(x):
    for crew_member in x:
        if crew_member['job'] == 'Director':
            return crew_member['name']
    return np.nan
````

````python
# Define the new director feature

df['director'] = df['crew'].apply(get_director)

# Print the directors of the first five movies
df['director'].head()
````

````python
# Returns the list top 3 elements or entire list; whichever is more.

def generate_list(x):
    if isinstance(x, list):
        names = [i['name'] for i in x]

        # Check if more than 3 elements exist. If yes, return only first three.
        # If no, return entire list.

        if len(names) > 3:
            names = names[:3]
        return names

    # Return empty list in case of missing/malformed data
    return []
````

````python
# Apply the generate_list function to cast and keywords
df['cast'] = df['cast'].apply(generate_list)
df['keywords'] = df['keywords'].apply(generate_list)
````

````python
# Only consider a maximum of 3 genres
df['genres'] = df['genres'].apply(lambda x: x[:3])
````

````python
# Print the new features of the first 5 movies along with title
df[['title', 'cast', 'director', 'keywords', 'genres']].head()
````

````python
# Function to sanitize data to prevent ambiguity. It removes spaces and converts to lowercase

def sanitize(x):
    if isinstance(x, list):
        # Strip spaces and converts to lowercase
        return [str.lower(i.replace(" ", "")) for i in x]
    else:
        # Check if director exists. If not, return empty string
        if isinstance(x, str):
            return str.lower(x.replace(" ", ""))
        else:
            return ''


# 1️⃣ Why we sanitize

# Sanitizing means cleaning the data to remove inconsistencies, like:

# Extra spaces: "Science Fiction " vs "Science Fiction"

# Capitalization differences: "Action" vs "action"

# Formatting issues in lists: "Christopher Nolan" vs " christopher nolan "

# If we don’t sanitize, the computer sees these as different values, even though they represent the same thing.

# 2️⃣ Example in a movie recommender

# Suppose you have genres like this:

# movie	genres
# Movie A	["Action", "Thriller"]
# Movie B	["action", "thriller "]
# Without sanitization:

# "Action" ≠ "action"

# "Thriller" ≠ "thriller "

# Cosine similarity or other matching algorithms will treat these movies as less similar than they really are.

# With sanitization:

# Convert everything to lowercase and remove spaces → ["action","thriller"] for both movies

# Now the algorithm correctly sees them as similar.

# 3️⃣ What happens if we don’t sanitize

# Reduced accuracy: Movies that are actually similar may not be recognized as such.

# Inconsistent results: Searching by cast, director, or genre may fail to match correctly.

# Duplicate entries: Slight variations in text may create “different” items that are really the same.

# 4️⃣ Summary
# Before sanitize	After sanitize	Problem solved
# "Action"	"action"	Case sensitivity
# "Science Fiction "	"sciencefiction"	Extra spaces
# "Christopher Nolan"	"christophernolan"	Uniform text for matching

# ✅ In short:
# Sanitization ensures the recommender system can correctly compare movies. Without it, the algorithm may underestimate similarity or give wrong recommendations.
````

````python
# Apply the sanitize function to cast, keywords, director and genres
for feature in ['cast', 'director', 'genres', 'keywords']:
    df[feature] = df[feature].apply(sanitize)
````


### 1️⃣ The function:

```python
# Function that creates a soup out of the desired metadata
def create_soup(x):
    return ' '.join(x['keywords']) + ' ' + ' '.join(x['cast']) + ' ' + x['director'] + ' ' + ' '.join(x['genres'])
```

---

### 2️⃣ What “soup” means here

* **This `soup` is not BeautifulSoup**.
* In this context, “soup” is just a **term used in NLP/data processing**: it means **a single string that combines multiple pieces of text data**.
* Basically, we’re “mixing all metadata into one text blob” so that we can later convert it into a **vector for similarity calculations** (like TF-IDF or CountVectorizer).

---

### 3️⃣ Step-by-step explanation

* `x['keywords']` → a list of keywords

  * `' '.join(x['keywords'])` → joins them into a single string separated by spaces
* `x['cast']` → a list of cast member names

  * `' '.join(x['cast'])` → joins them into a single string
* `x['director']` → a single string (director name)
* `x['genres']` → a list of genres

  * `' '.join(x['genres'])` → joins them into a single string

Then all these strings are concatenated with `' '` (spaces) between them.

---

### 4️⃣ Example

Suppose a movie row looks like this:

```python
x = {
    'keywords': ['dream', 'subconscious'],
    'cast': ['Leonardo DiCaprio', 'Joseph Gordon-Levitt'],
    'director': 'Christopher Nolan',
    'genres': ['Action', 'Sci-Fi']
}
```

Calling:

```python
create_soup(x)
```

Returns:

```
'dream subconscious Leonardo DiCaprio Joseph Gordon-Levitt Christopher Nolan Action Sci-Fi'
```

✅ Now you have **one big string** representing all relevant metadata for that movie.

---

### 5️⃣ Why do this?

* To feed it into **vectorizers** like `CountVectorizer` or `TfidfVectorizer`
* This allows the recommender system to **compute similarity** between movies based on all metadata at once.

---

### 🧠 Analogy

Think of each movie’s metadata as ingredients:

* keywords → sugar
* cast → flour
* director → eggs
* genres → chocolate

`create_soup` **mixes them all together** into one “cake batter” (soup) that the similarity algorithm can taste (analyze).

---

So in this context, **“soup” just means a single combined string**.




````python
# Function that creates a soup out of the desired metadata

def create_soup(x):
    return ' '.join(x['keywords']) + ' ' + ' '.join(x['cast']) + ' ' + x['director'] + ' ' + ' '.join(x['genres'])
````

````python
# Create the new soup feature
df['soup'] = df.apply(create_soup, axis=1)
````

````python
# Display the soup of the first movie

df.iloc[0]['soup']
````

````python
# Import CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer

# Define a new CountVecotrizer object and create vectors for the soup
count = CountVectorizer(stop_words='english')
count_matrix = count.fit_transform(df['soup'])
````

````python
# Import cosine_similarity function
from sklearn.metrics.pairwise import cosine_similarity

# Compute the cosine similarity score(equivalent to dot product for tf-idf vectors)
cosine_sim2 = cosine_similarity(count_matrix, count_matrix)
````

````python
# Reset index of your df and construct reverse mapping again
df = df.reset_index()
indices2 = pd.Series(df.index, index=df['title'])
````

````python
# Function that takes in movie title as input and gives recommendations
def content_recommender(title, cosine_sim=cosine_sim2, df=df, indices=indices2):
    # Obtain the index of the movie that matches the title
    idx = indices[title]

    # Get the pairwsie similarity scores of all movies with that movie
    # And convert it into a list of tuples as described above
    sim_scores = list(enumerate(cosine_sim[idx]))

    # Sort the movies based on the cosine similarity scores
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Get the scores of the 10 most similar movies. Ignore the first movie.
    sim_scores = sim_scores[1:11]

    # Get the movie indices
    movie_indices = [i[0] for i in sim_scores]

    # Return the top 10 most similar movies
    return df['title'].iloc[movie_indices]
````

````python
content_recommender('The Lion King', cosine_sim2, df, indices2)
````
